# Spatial Omics Analysis Tutorial
This notebook demonstrates spatial transcriptomics analysis using Scanpy, Squidpy, and interactive visualizations.

Features:
- Load example Visium dataset
- Preprocessing and normalization (Seurat v3 flavor)
- PCA, neighbors, UMAP, and Leiden clustering
- Static spatial plots
- Interactive plots with Plotly and Napari

**💡 5-Year-Old Analogy:** Imagine we are mapping a city full of houses, each with little lights (genes). We'll color the houses by their light patterns to find neighborhoods and see which lights are on.

## 1️⃣ Load Example Dataset
- We will use a mouse brain Visium dataset.
- `adata` stores our data: each row is a spot (cell), each column is a gene.

**5-Year-Old Analogy:** Each house in the city has a list of lights. `adata` is our city map showing every house and its lights.

## 2️⃣ Preprocessing
- Normalize gene counts so cells are comparable.
- Log-transform data to reduce effect of very bright lights.
- Select highly variable genes to focus on interesting lights.
- Scale values so no single gene dominates.

**5-Year-Old Analogy:**
1. Make sure all houses have a fair number of lights.
2. Squish the brightness so huge lights don’t scare us.
3. Pick the lights that change a lot from house to house.
4. Make all lights fair size to compare easily.

## 3️⃣ PCA, neighbors, UMAP, and clustering
- `PCA` compresses thousands of genes into a few main patterns.
- `neighbors` finds similar cells in gene space.
- `UMAP` maps cells into 2D for visualization.
- `Leiden` clustering groups similar cells.

**5-Year-Old Analogy:**
We shrink all lights into a few main colors, find houses with similar lights, draw a map showing neighborhoods, and color houses in clusters.

## 4️⃣ Visualize UMAP
- Shows clusters of cells in 2D.

**5-Year-Old Analogy:** Map of city with neighborhoods colored differently.

## 5️⃣ Static Spatial Plots
- Plot actual tissue coordinates.
- Can color by cluster or by gene expression.

**5-Year-Old Analogy:** Put houses back on the real city map and color them by neighborhood or light patterns.

## 6️⃣ Neighborhood Graph & Enrichment
- Compute which cells are near each other.
- Check if clusters like to be neighbors more than random.
- Plot enrichment to see cluster proximity.

**5-Year-Old Analogy:** Find which neighborhoods like to live next to each other.

## 7️⃣ Interactive Plot with Plotly
- Scatter plot of spatial coordinates.
- Hover to see cluster and gene expression.

**5-Year-Old Analogy:** Magic map where you can click houses to see which lights are on.

## 8️⃣ Interactive Visualization with Napari (Optional)
- Add points for each cell, color by gene expression or cluster.
- Blend layers for better visualization.

**5-Year-Old Analogy:** Look at the city with a magical zoomable camera. You can see the lights and neighborhoods interactively.
